```python
x = 'elle'

# Longeur de la chaîne de charactère
charLen = len(x)
#print(charLen)

pointeurGaucheIdx = 0;
pointeurDroiteIdx = charLen -1;

isPalindrome = True;

while pointeurGaucheIdx < pointeurDroiteIdx:
    if (x[pointeurGaucheIdx] != x[pointeurDroiteIdx]):
        isPalindrome = False;
        break;
    else:
        pointeurGaucheIdx +=1;
        pointeurDroiteIdx -=1;

print(isPalindrome)
```
